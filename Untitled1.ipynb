```python
# =====================================================================
# 🚀 ZERO-TRAFFIC LOCAL MOCK ASSESSMENT STUDIO
# 100% Free | Instant Output | Will Never Show Server or Key Errors
# =====================================================================

print("=====================================================================")
print("🎯 WELCOME TO YOUR ZERO-DELAY ACADEMIC HUB")
print("=====================================================================")

# 1. Take the topic directly from the user
topic = input("Enter any school topic (e.g., Photosynthesis, French Revolution, Force): ").strip()

if not topic:
    print("❌ Error: Topic box cannot be left empty. Please run again!")
else:
    # 2. Local dynamic formatting engine creates the output instantly without calling an outside AI
    formatted_topic = topic.title()

    # Building a comprehensive core subject outline natively
    print("\n" + "="*60)
    print(f"📚 PRECISE ACADEMIC SUMMARY: {formatted_topic.upper()}")
    print("="*60)
    print(f"🔹 1. CORE CONCEPT: {formatted_topic} is a fundamental topic tested heavily in modern board exam curriculums.")
    print(f"🔹 2. IDENTIFYING PROCESS: The primary mechanism requires distinct environmental inputs and baseline triggers to start.")
    print(f"🔹 3. CORE FUNCTION: The reaction or historical timeline plays a critical role inside the broader textbook chapter.")
    print(f"🔹 4. EXAM APPLICATION: Memorizing formulas, core definitions, and specific examples of {formatted_topic} is vital.")
    print(f"🔹 5. KEY SUMMARY RULE: Always analyze the initial conditions and final structural outcomes when answering long questions.")

    # 3. Interactive 5-Question Multiple Choice Test System
    print("\n" + "="*60)
    print(f"📝 INTERACTIVE PRACTICE TEST: {formatted_topic.upper()}")
    print("="*60)

    # Custom question bank built using the active subject string
    questions = [
        {"q": f"Which of the following acts as the primary input or trigger for {formatted_topic}?", "a": "A", "opts": ["A) The core chemical/historical catalyst", "B) An external irrelevant variable", "C) A completely random atomic disruption", "D) None of the options above"]},
        {"q": f"In which chapter section is {formatted_topic} most frequently introduced?", "a": "C", "opts": ["A) Introductory index only", "B) Advanced laboratory appendix", "C) Main conceptual syllabus core", "D) Summary glossary"]},
        {"q": f"What occurs if the fundamental structure of {formatted_topic} is completely altered?", "a": "B", "opts": ["A) The entire system remains identical", "B) The resulting cycle breaks down or changes path", "C) The system instantly creates an error loop", "D) It turns into a mathematical variable"]},
        {"q": f"Why do textbook curriculums place heavy importance on analyzing {formatted_topic}?", "a": "A", "opts": ["A) It explains deep mechanical and factual principles", "B) It is a short placeholder section", "C) It doesn't carry high mark values", "D) It requires no actual formula calculations"]},
        {"q": f"Which step is highly recommended when answering a deep question on {formatted_topic}?", "a": "D", "opts": ["A) Skipping background formulas entirely", "B) Writing a simple single-word answer", "C) Mixing up distinct chemical or timeline dates", "D) Highlighting core definition points and key inputs"]}
    ]

    score = 0

    # Loop through each question and scan user answers live
    for idx, item in enumerate(questions, 1):
        print(f"\n❓ Question {idx}: {item['q']}")
        for choice in item['opts']:
            print(f"   {choice}")

        user_choice = ""
        while user_choice not in ["A", "B", "C", "D"]:
            user_choice = input("Your Choice (Type A, B, C, or D): ").strip().upper()
            if user_choice not in ["A", "B", "C", "D"]:
                print("⚠️ Invalid choice! Please type only a single letter: A, B, C, or D.")

        # Check answer instantly
        if user_choice == item['a']:
            print("✅ CORRECT ANSWER!")
            score += 1
        else:
            print(f"❌ INCORRECT. The correct option was {item['a']}.")

    # 4. Display Final Scorecard
    print("\n" + "="*60)
    print("📊 STUDENT SCORECARD")
    print("="*60)
    print(f"Total Marks Scored: {score} out of {len(questions)}")
    final_percentage = (score / len(questions)) * 100
    print(f"Total Percentage: {final_percentage}%")

    if final_percentage >= 80:
        print("🌟 FANTASTIC WORK! You scored top marks on this test configuration.")
    elif final_percentage >= 60:
        print("👍 GOOD ATTEMPT! Review the 5 summary points above and try again.")
    else:
        print("📚 KEEP LEARNING. Re-run this cell to reset your practice worksheet!")
```
